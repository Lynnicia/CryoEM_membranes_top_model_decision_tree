# Top Model Decision Tree

This is a model screening framework to test segmentation model performance of fine-tuned pre-trained models of YOLOv11, YOLO26, U-Net, Detectron2 and SAM3 to generate segmentation masks and metrics of predicted bacterial inner membranes (IM) and outer membranes (OM) in cryoEM images. Low-dose (LD) and ultralow-dose (ULD) cryoEM datasets of bacteria are included as examples in this notebook.

This notebook is hardcoded to screen images in the "Datasets" directory framework. To use a custom dataset, please delete the images and annotations in their respective folders (leaving folder names intact) and place your images and annotations with the exact same format.

````python
import seaborn as sns
import matplotlib.pyplot as plt

csv_path = "/content/CryoEM_membranes_top_model_decision_tree/top_model_table.csv"
df = pd.read_csv(csv_path)


dfs_ld = df[
    (df["Test_Electron_Dose"] == test_electron_dose_decision) &
    (df["Test_Image_Size"] == test_image_size_decision) &
    (df["Model_Electron_Dose"] == model_electron_dose_decision)
]
color_map = {
    "U-Net": "grey",
    "YOLOv11": "blue",
    "YOLO26": "purple",
    "Detectron2": "green",
    "SAM3": "orange"
}

g = sns.catplot(
    data=dfs_ld,
    kind="bar",
    x="Model",
    y="total_bacteria",
    hue="Model_Architecture",
    palette=color_map,
    errorbar=None,
    height=2,
    aspect=2,
    legend=False
)

for ax in g.axes.flat:
    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_color("black")
        spine.set_linewidth(1)

g.set_axis_labels("", "Total Bacteria")
ax.tick_params(axis='both', labelsize=8)

from matplotlib.ticker import AutoMinorLocator

for ax in g.axes.flat:
    ax.yaxis.set_minor_locator(AutoMinorLocator(4))  # 2 minor ticks between majors
    ax.tick_params(axis='x', which='minor', length=3)

for ax in g.axes.flat:
    ax.axhline(manual_count, color='black', linestyle='--', linewidth=1)
for ax in g.axes.flat:
    ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')
plt.show()
````

````python
import seaborn as sns
import matplotlib.pyplot as plt

csv_path = "/content/CryoEM_membranes_top_model_decision_tree/top_model_table.csv"
df = pd.read_csv(csv_path)


#====================================================

# Scatter Plot


df["Model_Image_Size"] = pd.to_numeric(
    df["Model_Image_Size"],
    errors="coerce"
)
df = df.dropna(subset=["Model_Image_Size"])

dfs = df.sort_values("FPS", ascending=False)

color_map = {
    "U-Net": "grey",
    "YOLOv11": "blue",
    "YOLO26": "purple",
    "Detectron2": "green",
    "SAM3": "orange"
}

def color_bars(ax, subset):
    model_y = {t.get_text(): pos for pos, t in zip(ax.get_yticks(), ax.get_yticklabels())}

    for patch in ax.patches:
        bar_y = patch.get_y() + patch.get_height() / 2

        if not model_y:
            continue
        closest_model = min(model_y, key=lambda m: abs(model_y[m] - bar_y))

        matched = subset[subset["Model"] == closest_model]
        if matched.empty:
            continue

        tick_y = model_y[closest_model]
        is_ld = bar_y < tick_y

        arch = matched["Model_Architecture"].iloc[0]
        color = color_map.get(arch, "black")

        if is_ld:
            patch.set_facecolor(color)
            patch.set_edgecolor(color)
        else:
            patch.set_facecolor("white")
            patch.set_edgecolor(color)
            patch.set_linewidth(1.5)

    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_color("black")
        spine.set_linewidth(1)

plt.rcParams["figure.dpi"] = 300

for size in [640, 1024]:
    subset = dfs[dfs["Test_Image_Size"] == size]

    g = sns.catplot(
        data=subset,
        kind="bar",
        y="Model",
        x="FPS",
        hue="Model_Electron_Dose",
        palette="light:gray",
        errorbar=None,
        height=2,
        aspect=2
    )  # paper used height = 5, aspect = 1

    ax = g.axes.flat[0]
    color_bars(ax, subset)
    ax.set_xlim(0, 140)
    g.legend.remove()
    ax.tick_params(axis='both', labelsize=8)

    g.set_axis_labels("FPS", "Model")
    ax.text(0.98, 0.02, f"{test_electron_dose_decision} Test Image Size: {size}",
        transform=ax.transAxes,
        ha="right", va="bottom",
        fontsize=8)
    plt.tight_layout()
    plt.show()
````
